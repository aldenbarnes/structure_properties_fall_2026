# SF rent lookup

Type an address (or paste an apartments.com URL), pick the unit, edit any feature, and get a rent estimate from `rent_model.py`.

* **Cost per lookup:** an apartments.com URL or listing key uses 1 RealtyAPI request (`/details/byid`). A street address uses 2: one search to find the listing, then `/details/byid` for the property you pick. Results are cached, so switching units or editing features is free.
* **Every box can be changed.** Blank means unknown; the model then uses its training average and lists that under Warnings.
* **Read the warnings.** The model was trained mostly on large downtown buildings (SoMa, Tenderloin, Mission Bay). Estimates for outer neighborhoods and 3+ bedrooms are rough.
* Store your key as a Colab secret named `REALTYAPI_KEY` (key icon in the left sidebar), or paste it at the prompt. Never put it in the notebook.
